```python
model = joblib.load("credit_risk_model_lgbm.pkl")
```

```python
class Borrower(BaseModel):
    loan_amount: float
    income: float
    Credit_Score: float
    LTV: float
    dtir1: float
```

```python
app = FastAPI(title="Credit Risk Scoring API")
```

```python
def expected_loss_engine(prob, loan_amount):

    expected_loss = prob * loan_amount

    if expected_loss < 50000:
        priority = "Low"
    elif expected_loss < 200000:
        priority = "Medium"
    elif expected_loss < 500000:
        priority = "High"
    else:
        priority = "Critical"

    return expected_loss, priority
```

```python
@app.post("/predict-risk")
def predict_risk(data: Borrower):

    input_df = pd.DataFrame([data.dict()])

    # Align columns with trained model
    input_df = input_df.reindex(columns=model.feature_name_, fill_value=0)

    prob = model.predict_proba(input_df)[0][1]

    expected_loss, priority = expected_loss_engine(prob, data.loan_amount)

    return {
        "default_probability": round(prob,2),
        "expected_loss": round(expected_loss,2),
        "priority": priority
    }
```

```python
def risk_scoring_engine(input_data):

    # You can replace this with your trained model later
    prob = 0.7

    loan_amount = input_data["loan_amount"]

    expected_loss = prob * loan_amount

    if expected_loss < 50000:
        priority = "Low"
        action = "Send reminder"
    elif expected_loss < 200000:
        priority = "Medium"
        action = "Call borrower"
    elif expected_loss < 500000:
        priority = "High"
        action = "Assign recovery officer"
    else:
        priority = "Critical"
        action = "Legal action"

    return {
        "default_probability": round(prob,2),
        "expected_loss": round(expected_loss,2),
        "priority": priority,
        "recommended_action": action
    }
```

```python
from fastapi import FastAPI
from pydantic import BaseModel

app = FastAPI(title="CreditPath AI Risk Scoring API")

class BorrowerInput(BaseModel):
    income: float
    credit_score: float
    debt_ratio: float
    loan_amount: float
    missed_payments: int

@app.post("/risk-score")
def get_risk_score(data: BorrowerInput):
    input_data = data.dict()
    result = risk_scoring_engine(input_data)
    return result
```
