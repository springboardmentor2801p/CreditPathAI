```python
def user_recommendation_engine(loan_type, income, credit_score, loan_amount, missed_payments):

    # Risk logic
    if credit_score > 750:
        risk = "Low"
    elif credit_score > 600:
        risk = "Medium"
    else:
        risk = "High"

    # Advice
    if risk == "Low":
        advice = "You are eligible. Maintain your credit score."
        tips = [
            "Maintain timely EMI payments",
            "Keep credit utilization low"
        ]

    elif risk == "Medium":
        advice = "Improve credit score and reduce debt before applying."
        tips = [
            "Pay EMIs on time",
            "Avoid multiple loans",
            "Reduce credit card usage"
        ]

    else:
        advice = "High risk. Avoid applying now."
        tips = [
            "Clear existing debts",
            "Avoid applying for new loans",
            "Improve repayment history"
        ]

    # Loan type suggestion
    if loan_type == "home":
        suggestion = "Long-term planning required"
    elif loan_type == "car":
        suggestion = "Short-term manageable loan"
    else:
        suggestion = "Check loan terms carefully"

    # Risk logic
    if credit_score > 750:
        risk = "Low"
    elif credit_score > 600:
        risk = "Medium"
    else:
        risk = "High"

    # Advice
    if risk == "Low":
        advice = "You are eligible. Maintain your credit score."
        tips = [
            "Maintain timely EMI payments",
            "Keep credit utilization low"
        ]

    elif risk == "Medium":
        advice = "Improve credit score and reduce debt before applying."
        tips = [
            "Pay EMIs on time",
            "Avoid multiple loans",
            "Reduce credit card usage"
        ]

    else:
        advice = "High risk. Avoid applying now."
        tips = [
            "Clear existing debts",
            "Avoid applying for new loans",
            "Improve repayment history"
        ]

    # Loan type suggestion
    if loan_type == "home":
        suggestion = "Long-term planning required"
    elif loan_type == "car":
        suggestion = "Short-term manageable loan"
    else:
        suggestion = "Check loan terms carefully"

    return {
    "risk_level": risk,
    "recommendation_summary": [
        advice,
        f"Loan Insight: {suggestion}",
        "Action: " + tips[0]   # shows one key actionable tip
    ],
    "tips": tips
  }
```

```python
import random
import json

loan_types = ["home", "car", "personal"]

loan_type = random.choice(loan_types)
income = random.randint(20000, 100000)
credit_score = random.randint(500, 800)
loan_amount = random.randint(50000, 500000)
missed_payments = random.randint(0, 5)

result_user = user_recommendation_engine(
    loan_type=loan_type,
    income=income,
    credit_score=credit_score,
    loan_amount=loan_amount,
    missed_payments=missed_payments
)

print("Input Values:")
print(f"Loan Type: {loan_type}")
print(f"Income: {income}")
print(f"Credit Score: {credit_score}")
print(f"Loan Amount: {loan_amount}")
print(f"Missed Payments: {missed_payments}")

print("\nOutput:")
print(json.dumps(result_user, indent=4))
```
