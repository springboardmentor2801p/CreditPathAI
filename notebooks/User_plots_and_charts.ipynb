```python
import matplotlib.pyplot as plt
import random
import json
```

```python
def user_recommendation_engine(loan_type, income, credit_score, loan_amount, missed_payments):

    # Risk logic
    if credit_score > 750:
        risk = "Low"
    elif credit_score > 600:
        risk = "Medium"
    else:
        risk = "High"

    # Advice
    if risk == "Low":
        advice = "You are eligible. Maintain your credit score."
        tips = [
            "Maintain timely EMI payments",
            "Keep credit utilization low"
        ]

    elif risk == "Medium":
        advice = "Improve credit score and reduce debt before applying."
        tips = [
            "Pay EMIs on time",
            "Avoid multiple loans",
            "Reduce credit card usage"
        ]

    else:
        advice = "High risk. Avoid applying now."
        tips = [
            "Clear existing debts",
            "Avoid applying for new loans",
            "Improve repayment history"
        ]

    # Loan type suggestion
    if loan_type == "home":
        suggestion = "Long-term planning required"
    elif loan_type == "car":
        suggestion = "Short-term manageable loan"
    else:
        suggestion = "Check loan terms carefully"

    return {
        "risk_level": risk,
        "user_advice": advice,
        "loan_type_suggestion": suggestion,
        "tips": tips   # ✅ NEW ADDED
    }
```

```python
loan_type = random.choice(["home", "car", "personal"])
income = random.randint(20000, 100000)
credit_score = random.randint(500, 800)
loan_amount = random.randint(50000, 500000)
missed_payments = random.randint(0, 5)

result_user = user_recommendation_engine(
    loan_type, income, credit_score, loan_amount, missed_payments
)

print(json.dumps(result_user, indent=4))
```

```python
risk = result_user["risk_level"]
```

```python
#convert risk to numeric
risk_map = {"Low": 1, "Medium": 2, "High": 3}
risk_value = risk_map[risk]
```

```python
plt.figure()
plt.bar(["Risk Level"], [risk_value])
plt.title(f"User Risk Level ({risk})")
plt.show()
```

```python
plt.figure()
plt.bar(["Income", "Loan Amount"], [income, loan_amount])
plt.title("Income vs Loan Amount")
plt.show()
```

```python
# ---- BAR CHART (USER VIEW) ----
plt.figure()
plt.bar(["Credit Score"], [credit_score])
plt.title("Your Credit Score")
plt.ylabel("Score")
plt.show()
```

```python
# ---- CREDIT SCORE CATEGORY (USER VIEW) ----
if credit_score > 750:
    values = [credit_score, 900 - credit_score]
    labels = ["Excellent Score", "Scope for Improvement"]
elif credit_score > 600:
    values = [credit_score, 900 - credit_score]
    labels = ["Average Score", "Improve for Better Eligibility"]
else:
    values = [credit_score, 900 - credit_score]
    labels = ["Low Score", "High Improvement Needed"]

plt.figure()
plt.pie(values, labels=labels, autopct="%1.1f%%")
plt.title("Your Credit Score Distribution")
plt.show()
```
